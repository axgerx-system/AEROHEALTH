## 1. Dataset Loading and Initial Validation / Chargement et validation initiale du jeu de données

### EN
We begin with NASA C-MAPSS FD001. Each observation contains an engine identifier, operating cycle, three operating settings, and 21 sensor measurements. Training trajectories run to simulated failure. Test trajectories are truncated and are paired with official terminal RUL targets in `RUL_FD001.txt`.

### FR
Nous commençons avec le sous-ensemble NASA C-MAPSS FD001. Chaque observation contient un identifiant moteur, un cycle de fonctionnement, trois paramètres de fonctionnement et 21 mesures de capteurs. Les trajectoires d'entraînement vont jusqu'à la défaillance simulée. Les trajectoires de test sont tronquées et associées aux cibles RUL terminales officielles du fichier `RUL_FD001.txt`.


```python
# EN: Import the scientific stack and resolve the project root before loading data.
# FR: Importer les bibliothèques scientifiques et résoudre la racine du projet avant de charger les données.

from pathlib import Path
import os

# EN: Keep native numerical libraries single-threaded for reproducible, stable notebook execution.
# FR: Garder les bibliothèques numériques natives sur un seul thread pour une exécution stable et reproductible.
os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# EN: Resolve the data directory without relying on a user-specific absolute path.
# FR: Résoudre le dossier des données sans dépendre d'un chemin absolu propre à un utilisateur.

data_candidates = [
    Path.cwd() / "data" / "raw",
    Path.cwd().parent / "data" / "raw",
    Path("/mnt/data/data/raw"),
    Path("/mnt/data/aerohealth_fixture_data"),
]

DATA_DIR = next(
    (candidate for candidate in data_candidates
     if all((candidate / name).exists() for name in [
         "train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt"
     ])),
    None,
)

if DATA_DIR is None:
    raise FileNotFoundError(
        "FD001 data files could not be resolved. Expected train_FD001.txt, "
        "test_FD001.txt, and RUL_FD001.txt in a project data/raw directory "
        "or the supplied /mnt/data/aerohealth_fixture_data directory."
    )

PROJECT_ROOT = DATA_DIR.parent.parent if DATA_DIR.name == "raw" else DATA_DIR
FIGURE_DIR = PROJECT_ROOT / "figures"
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# EN: Define the Aerohealth visual identity before any figure is created.
# FR: Définir l’identité visuelle Aerohealth avant de créer toute figure.
AERO_BG = "#0A1118"
AERO_CYAN = "#22B8CF"
AERO_STEEL = "#4C78A8"
AERO_ICE = "#D9F3F6"
AERO_TEXT = "#E8EEF2"
AERO_MUTED = "#7C8B96"
AERO_GRID = "#26343F"
AERO_WARNING = "#F5B942"
AERO_CRITICAL = "#E05260"
AERO_STABLE = "#39D98A"

AERO_COLORS = {
    "background": AERO_BG,
    "primary": AERO_CYAN,
    "secondary": AERO_STEEL,
    "highlight": AERO_ICE,
    "text": AERO_TEXT,
    "muted": AERO_MUTED,
    "grid": AERO_GRID,
    "warning": AERO_WARNING,
    "critical": AERO_CRITICAL,
    "stable": AERO_STABLE,
}

plt.rcParams.update({
    "figure.facecolor": AERO_BG,
    "axes.facecolor": AERO_BG,
    "axes.edgecolor": AERO_GRID,
    "axes.labelcolor": AERO_TEXT,
    "axes.titlecolor": AERO_TEXT,
    "text.color": AERO_TEXT,
    "xtick.color": AERO_MUTED,
    "ytick.color": AERO_MUTED,
    "grid.color": AERO_GRID,
    "grid.alpha": 0.55,
    "axes.titleweight": "bold",
    "font.size": 11,
    "axes.prop_cycle": plt.cycler(color=[AERO_CYAN, AERO_STEEL, AERO_WARNING, AERO_STABLE, AERO_ICE]),
})

AERO_CORR_CMAP = LinearSegmentedColormap.from_list(
    "aerohealth_correlation",
    [AERO_STEEL, AERO_BG, AERO_CRITICAL],
    N=256,
)

print("Resolved project/data root:", PROJECT_ROOT)
print("Data directory:", DATA_DIR)
print("Figure directory:", FIGURE_DIR)
print("Aerohealth visual identity loaded.")
```

```python
# EN: Load the official FD001 training, test, and terminal-RUL files.
# FR: Charger les fichiers officiels FD001 d’entraînement, de test et de RUL terminale.

COLUMNS = (
    ["unit_id", "cycle"]
    + [f"setting_{i}" for i in range(1, 4)]
    + [f"sensor_{i}" for i in range(1, 22)]
)

train_df = pd.read_csv(
    DATA_DIR / "train_FD001.txt",
    sep=r"\s+",
    header=None,
    names=COLUMNS,
)

test_df = pd.read_csv(
    DATA_DIR / "test_FD001.txt",
    sep=r"\s+",
    header=None,
    names=COLUMNS,
)

rul_df = pd.read_csv(
    DATA_DIR / "RUL_FD001.txt",
    sep=r"\s+",
    header=None,
    names=["RUL"],
)

print(f"Training data: {train_df.shape}")
print(f"Test data:     {test_df.shape}")
print(f"RUL targets:   {rul_df.shape}")
```

```python
# EN: Define sensor groups once before all EDA and modeling steps.
# FR: Définir une seule fois les groupes de capteurs avant toute l'EDA et la modélisation.

sensor_columns = [f"sensor_{i}" for i in range(1, 22)]

variable_sensors = [
    sensor
    for sensor in sensor_columns
    if train_df[sensor].nunique(dropna=False) > 1
]

constant_sensors = [
    sensor
    for sensor in sensor_columns
    if sensor not in variable_sensors
]

# EN: Validate the expected FD001 sensor structure.
# FR: Vérifier la structure attendue des capteurs pour FD001.

assert len(sensor_columns) == 21
assert len(variable_sensors) == 15
assert len(constant_sensors) == 6

print("Variable sensors:", len(variable_sensors))
print("Constant sensors:", constant_sensors)
```

```python
print("=== MISSING VALUES ===")
print(train_df.isna().sum().sum())

print("\n=== DUPLICATE ROWS ===")
print(train_df.duplicated().sum())

print("\n=== UNIT ID RANGE ===")
print(train_df["unit_id"].min(), "to", train_df["unit_id"].max())

print("\n=== CYCLE RANGE ===")
print(train_df["cycle"].min(), "to", train_df["cycle"].max())

print("\n=== UNIQUE VALUES PER FEATURE ===")
feature_columns = COLUMNS[2:]

unique_counts = train_df[feature_columns].nunique().sort_values()

print(unique_counts)
```

```python
engine_lifetimes = (
    train_df.groupby("unit_id")["cycle"]
    .max()
)

print("=== ENGINE LIFETIME STATISTICS ===")
print(engine_lifetimes.describe())

print("\n=== SHORTEST ENGINES ===")
print(engine_lifetimes.nsmallest(10))

print("\n=== LONGEST ENGINES ===")
print(engine_lifetimes.nlargest(10))
```

```python
# EDA STEP 2B  -  ENGINE LIFETIME DISTRIBUTION
# EN: Visualize the distribution of complete training-engine lifetimes.
# FR: Visualiser la distribution des durées de vie complètes des moteurs d'entraînement.

fig, ax = plt.subplots(figsize=(10, 5), facecolor=AERO_BG)
ax.set_facecolor(AERO_BG)

ax.hist(
    engine_lifetimes,
    bins=15,
    color=AERO_CYAN,
    edgecolor=AERO_BG,
    linewidth=0.8,
    alpha=0.95,
)

median_lifetime = engine_lifetimes.median()

ax.axvline(
    median_lifetime,
    color=AERO_ICE,
    linestyle="--",
    linewidth=1.8,
    label=f"Median / Médiane = {median_lifetime:.0f} cycles",
)

ax.set_title(
    "Training Engine Lifetime Distribution / Distribution de la durée de vie des moteurs",
    fontsize=14,
    fontweight="bold",
    color=AERO_TEXT,
)

ax.set_xlabel(
    "Engine lifetime (cycles) / Durée de vie du moteur (cycles)",
    color=AERO_TEXT,
)

ax.set_ylabel(
    "Number of engines / Nombre de moteurs",
    color=AERO_TEXT,
)

ax.grid(True, color=AERO_GRID, alpha=0.55, linewidth=0.7)
ax.tick_params(colors=AERO_MUTED)

legend = ax.legend(
    facecolor=AERO_BG,
    edgecolor=AERO_GRID,
)
for text in legend.get_texts():
    text.set_color(AERO_TEXT)

plt.tight_layout()
plt.show()
plt.close("all")
```

```python
# ============================================================
# EDA STEP 7C  -  SENSOR TEMPORAL ASSOCIATION
# ============================================================

from scipy.stats import spearmanr
import pandas as pd

# EN: Measure the monotonic relationship between each sensor
#     and relative engine life for each engine.
# FR: Mesurer la relation monotone entre chaque capteur
#     et la durée de vie relative pour chaque moteur.

sensor_trend_results = []

for sensor in variable_sensors:

    engine_correlations = []

    for unit_id, engine_data in train_df.groupby("unit_id"):

        sensor_values = engine_data[sensor]

        # EN: Skip constant sensor trajectories because Spearman
        #     correlation is undefined when one input is constant.
        # FR: Ignorer les trajectoires de capteur constantes car la
        #     corrélation de Spearman est indéfinie lorsqu'une entrée est constante.

        if sensor_values.nunique() <= 1:
            continue

        r, _ = spearmanr(
            engine_data["relative_life"],
            sensor_values
        )

        if pd.notna(r):
            engine_correlations.append(r)

    correlations = pd.Series(engine_correlations)

    sensor_trend_results.append({
        "sensor": sensor,
        "median_spearman": correlations.median(),
        "mean_spearman": correlations.mean(),
        "std_spearman": correlations.std(),
        "positive_engines_pct": (
            (correlations > 0).mean() * 100
        ),
        "engines": len(correlations)
    })

sensor_trend_summary = (
    pd.DataFrame(sensor_trend_results)
    .sort_values(
        "median_spearman",
        ascending=False
    )
    .reset_index(drop=True)
)

print("Sensor temporal association with relative engine life:")
display(
    sensor_trend_summary.round(3)
)
```

```python
# STEP 9A - Sensor ↔ RUL relationship
# EN: Measure the monotonic relationship between each sensor and the actual RUL target within each engine.
# FR: Mesurer la relation monotone entre chaque capteur et la cible RUL réelle au sein de chaque moteur.

import numpy as np
from scipy.stats import spearmanr

sensor_rul_results = []

for sensor in variable_sensors:
    engine_correlations = []

    for unit_id, engine_data in train_df.groupby("unit_id"):
        sensor_values = engine_data[sensor]
        rul_values = engine_data["RUL"]

        # EN: Skip sensors that are constant within an engine because Spearman correlation is undefined.
        # FR: Ignorer les capteurs constants au sein d'un moteur car la corrélation de Spearman est indéfinie.
        if sensor_values.nunique() <= 1:
            continue

        correlation, _ = spearmanr(sensor_values, rul_values)

        if not np.isnan(correlation):
            engine_correlations.append(correlation)

    if engine_correlations:
        sensor_rul_results.append({
            "sensor": sensor,
            "median_spearman": np.median(engine_correlations),
            "mean_spearman": np.mean(engine_correlations),
            "std_spearman": np.std(engine_correlations),
            "negative_engines_pct": (
                np.mean(np.array(engine_correlations) < 0) * 100
            ),
            "positive_engines_pct": (
                np.mean(np.array(engine_correlations) > 0) * 100
            ),
            "engines": len(engine_correlations)
        })

sensor_rul_df = (
    pd.DataFrame(sensor_rul_results)
    .sort_values("median_spearman")
    .reset_index(drop=True)
)

sensor_rul_df
```

```python
# STEP 11  -  SENSOR ↔ RUL RELATIONSHIP
# EN: Visualize median per-engine Spearman association between sensors and RUL.
# FR: Visualiser l'association médiane de Spearman entre les capteurs et la RUL.

plot_df = sensor_rul_df.sort_values("median_spearman")

fig, ax = plt.subplots(
    figsize=(10, 7),
    facecolor=AERO_BG,
)

ax.set_facecolor(AERO_BG)

ax.barh(
    plot_df["sensor"],
    plot_df["median_spearman"],
    color=AERO_CYAN,
)

ax.axvline(
    0,
    color=AERO_ICE,
    linewidth=1.2,
)

ax.set_title(
    "Sensor ↔ RUL Relationship  -  Median Spearman Correlation /\n"
    "Relation capteur ↔ RUL  -  Corrélation médiane de Spearman",
    color=AERO_TEXT,
    fontsize=14,
    pad=14,
)

ax.set_xlabel(
    "Median Spearman correlation with RUL / "
    "Corrélation médiane de Spearman avec la RUL",
    color=AERO_TEXT,
)

ax.set_ylabel(
    "Sensor / Capteur",
    color=AERO_TEXT,
)

ax.tick_params(axis="both", colors=AERO_MUTED)

ax.grid(
    axis="x",
    color=AERO_GRID,
    linewidth=0.8,
    alpha=0.8,
)

plt.tight_layout()
plt.show()
plt.close("all")
```

```python
# STEP 9C - Sensor redundancy among RUL-associated sensors
# EN: Identify strongly correlated sensor pairs among sensors associated with RUL.
# FR: Identifier les paires de capteurs fortement corrélées parmi les capteurs associés à la RUL.

# EN: Select sensors with an absolute median Spearman relationship with RUL >= 0.60.
# FR: Sélectionner les capteurs dont la relation médiane absolue avec la RUL est >= 0.60.
rul_relevant_sensors = sensor_rul_df[
    sensor_rul_df["median_spearman"].abs() >= 0.60
]["sensor"].tolist()

sensor_corr = train_df[rul_relevant_sensors].corr(method="pearson")

redundancy_results = []

for i, sensor_a in enumerate(rul_relevant_sensors):
    for sensor_b in rul_relevant_sensors[i + 1:]:
        correlation = sensor_corr.loc[sensor_a, sensor_b]

        redundancy_results.append({
            "sensor_a": sensor_a,
            "sensor_b": sensor_b,
            "pearson_correlation": correlation,
            "absolute_correlation": abs(correlation)
        })

redundancy_df = (
    pd.DataFrame(redundancy_results)
    .sort_values("absolute_correlation", ascending=False)
    .reset_index(drop=True)
)

redundancy_df.head(15)
```

```python
# STEP 9D - Consolidate sensor-RUL evidence
# EN: Combine RUL association and sensor redundancy evidence into one summary table.
# FR: Combiner les relations avec la RUL et les informations de redondance dans un tableau de synthèse.

sensor_summary = sensor_rul_df[
    [
        "sensor",
        "median_spearman",
        "mean_spearman",
        "std_spearman",
        "negative_engines_pct",
        "positive_engines_pct",
        "engines"
    ]
].copy()

# EN: Calculate the strongest absolute correlation with another RUL-relevant sensor.
# FR: Calculer la plus forte corrélation absolue avec un autre capteur pertinent pour la RUL.
max_redundancy = []

for sensor in sensor_summary["sensor"]:
    other_correlations = redundancy_df[
        (redundancy_df["sensor_a"] == sensor) |
        (redundancy_df["sensor_b"] == sensor)
    ]["absolute_correlation"]

    max_redundancy.append(
        other_correlations.max() if len(other_correlations) > 0 else np.nan
    )

sensor_summary["max_sensor_correlation"] = max_redundancy

sensor_summary = sensor_summary.sort_values(
    "median_spearman",
    key=lambda x: x.abs(),
    ascending=False
).reset_index(drop=True)

sensor_summary
```

```python
# STEP 10A - Restore missing causal temporal feature variables
# EN: Use the existing training state and create only the missing temporal-feature objects.
# FR: Utiliser l'état d'entraînement existant et créer uniquement les objets temporels manquants.

assert len(sensor_columns) == 21
assert len(variable_sensors) == 15

feature_df = (
    train_df
    .sort_values(["unit_id", "cycle"])
    .reset_index(drop=True)
)

for sensor in variable_sensors:

    feature_df[f"{sensor}_rollmean_5"] = (
        feature_df.groupby("unit_id")[sensor]
        .transform(
            lambda x: x.rolling(
                window=5,
                min_periods=1
            ).mean()
        )
    )

    feature_df[f"{sensor}_rollstd_5"] = (
        feature_df.groupby("unit_id")[sensor]
        .transform(
            lambda x: x.rolling(
                window=5,
                min_periods=2
            ).std()
        )
    )

    feature_df[f"{sensor}_diff_1"] = (
        feature_df.groupby("unit_id")[sensor]
        .diff()
    )

print("feature_df shape:", feature_df.shape)
print("Variable sensors:", len(variable_sensors))
print("RUL present:", "RUL" in feature_df.columns)
print(
    "Missing values introduced:",
    int(feature_df.isna().sum().sum())
)
```

```python
# STEP 10B - Engine cycle position
# EN: Add the current engine cycle as a causal indicator of engine operating age.
# FR: Ajouter le cycle courant du moteur comme indicateur causal de son âge de fonctionnement.

# EN: Copy the engineered dataset before adding the new feature.
# FR: Copier le jeu de données avec variables temporelles avant d'ajouter la nouvelle variable.
feature_df = feature_df.copy()

# EN: Cycle number represents the current observed operating age of each engine.
# FR: Le numéro de cycle représente l'âge de fonctionnement actuellement observé du moteur.
feature_df["cycle_age"] = feature_df["cycle"]

print("Feature dataset shape:", feature_df.shape)

print("\nNew feature:")
print(feature_df[["unit_id", "cycle", "cycle_age", "RUL"]].head(10))
```

```python
# STEP 10C - Add 10-cycle causal trend features
# EN: Compute backward-looking 10-cycle linear trends without using future observations.
# FR: Calculer les tendances linéaires rétrospectives sur 10 cycles sans utiliser d'observations futures.

# EN: Use an algebraically equivalent least-squares slope formula with convolution.
#     This preserves the original feature definition while avoiding a slow rolling np.polyfit loop.
# FR: Utiliser une formule de pente des moindres carrés algébriquement équivalente avec convolution.
#     La définition originale de la variable est conservée, mais la boucle np.polyfit lente est évitée.

def causal_rolling_slope(values, window=10):
    arr = np.asarray(values, dtype=float)
    result = np.full(arr.shape, np.nan, dtype=float)

    if len(arr) < 2:
        return result

    # EN: Handle the short initial windows explicitly.
    # FR: Traiter explicitement les premières fenêtres courtes.
    for n in range(2, min(window, len(arr) + 1)):
        y = arr[:n]
        if np.isnan(y).any():
            continue
        x = np.arange(n, dtype=float)
        centered_x = x - x.mean()
        denominator = np.sum(centered_x ** 2)
        result[n - 1] = np.sum(centered_x * (y - y.mean())) / denominator

    if len(arr) >= window:
        x = np.arange(window, dtype=float)
        centered_x = x - x.mean()
        denominator = np.sum(centered_x ** 2)
        weights = centered_x[::-1]
        numerator = np.convolve(arr, weights, mode="valid")
        result[window - 1:] = numerator / denominator

    return result

for sensor in variable_sensors:
    feature_df[f"{sensor}_trend_10"] = (
        feature_df.groupby("unit_id")[sensor]
        .transform(lambda x: causal_rolling_slope(x.to_numpy(), window=10))
    )

print("feature_df shape:", feature_df.shape)

trend_columns = [
    col for col in feature_df.columns
    if col.endswith("_trend_10")
]

print("Trend features:", len(trend_columns))
print("Expected variable sensors:", len(variable_sensors))
print(
    "Trend missing values:",
    int(feature_df[trend_columns].isna().sum().sum())
)
```

```python
# STEP 10D - Freeze the 78-feature model representation
# EN: Build the exact frozen feature list without including exploratory variables.
# FR: Construire la liste exacte des 78 variables sans inclure les variables exploratoires.

# EN: Constant FD001 sensors excluded from the predictive representation.
# FR: Capteurs constants de FD001 exclus de la représentation prédictive.

# EN: Retain only the 15 variable raw sensors.
# FR: Conserver uniquement les 15 capteurs bruts variables.

raw_model_sensors = [
    sensor
    for sensor in sensor_columns
    if sensor not in constant_sensors
]

# EN: Explicitly construct the frozen 78-feature representation.
# FR: Construire explicitement la représentation figée de 78 variables.

model_feature_columns = (
    ["setting_1", "setting_2"]
    + raw_model_sensors
    + [f"{sensor}_rollmean_5" for sensor in variable_sensors]
    + [f"{sensor}_rollstd_5" for sensor in variable_sensors]
    + [f"{sensor}_diff_1" for sensor in variable_sensors]
    + ["cycle_age"]
    + [f"{sensor}_trend_10" for sensor in variable_sensors]
)

# EN: Validate the exact frozen representation.
# FR: Vérifier exactement la représentation figée.

assert len(model_feature_columns) == 78
assert len(set(model_feature_columns)) == 78

assert "relative_life" not in model_feature_columns

assert len(raw_model_sensors) == 15
assert len(variable_sensors) == 15
assert len(constant_sensors) == 6
assert set(constant_sensors) == {"sensor_1", "sensor_5", "sensor_10", "sensor_16", "sensor_18", "sensor_19"}

# EN: Verify that every model feature exists in the training feature table.
# FR: Vérifier que chaque variable du modèle existe dans la table d'entraînement.

missing_from_training = [
    feature
    for feature in model_feature_columns
    if feature not in feature_df.columns
]

if missing_from_training:
    raise ValueError(
        f"Model features missing from training feature table: "
        f"{missing_from_training}"
    )

print("Model features:", len(model_feature_columns))
print("Expected:", 78)
print("Constant sensors excluded:", len(constant_sensors))
print("Raw sensors retained:", len(raw_model_sensors))
print("Variable sensors:", len(variable_sensors))
print("relative_life included:", "relative_life" in model_feature_columns)
print(
    "All model features present:",
    all(
        feature in feature_df.columns
        for feature in model_feature_columns
    )
)
```

```python
# STEP 11E - Lock the frozen 78-feature representation for modeling
# EN: Use the authoritative 78-feature list defined in STEP 10D.
# FR: Utiliser la liste autoritaire de 78 variables définie à l'étape 10D.

# EN: Candidate features are retained for auditability, but the model input is not redefined here.
# FR: Les variables candidates sont conservées pour la traçabilité, mais l'entrée du modèle n'est pas redéfinie ici.

assert len(model_feature_columns) == 78
assert len(set(model_feature_columns)) == 78
assert "setting_3" not in model_feature_columns
assert "relative_life" not in model_feature_columns

missing_from_model_table = [
    feature for feature in model_feature_columns
    if feature not in feature_df.columns
]
if missing_from_model_table:
    raise ValueError(
        f"Frozen model features missing from training feature table: {missing_from_model_table}"
    )

X_train_model = train_split[model_feature_columns].copy()
X_val_model = val_split[model_feature_columns].copy()

print("Candidate feature count before freeze:", len(feature_columns))
print("Frozen model feature count:", len(model_feature_columns))
print("X_train_model shape:", X_train_model.shape)
print("X_val_model shape:", X_val_model.shape)
print("setting_3 included:", "setting_3" in model_feature_columns)
print("relative_life included:", "relative_life" in model_feature_columns)
print("Missing values - training:", int(X_train_model.isna().sum().sum()))
print("Missing values - validation:", int(X_val_model.isna().sum().sum()))
```

```python
# STEP 12A - XGBoost baseline RUL model
# EN: Train a reproducible baseline XGBoost regression model without hyperparameter optimization.
# FR: Entraîner un modèle de régression XGBoost de référence reproductible sans optimisation des hyperparamètres.

from xgboost import XGBRegressor

# EN: Define a deliberately simple baseline configuration.
# FR: Définir une configuration volontairement simple pour le modèle de référence.
baseline_model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=1,
    tree_method="hist"
)

# EN: Train only on the training engines.
# FR: Entraîner uniquement sur les moteurs du jeu d'entraînement.
baseline_model.fit(
    X_train_model,
    y_train
)

print("Baseline XGBoost training complete.")
```

```python
# STEP 12B - Evaluate XGBoost baseline
# EN: Evaluate the baseline model on completely unseen validation engines.
# FR: Évaluer le modèle de référence sur des moteurs de validation complètement inconnus.

from sklearn.metrics import mean_absolute_error, mean_squared_error

# EN: Generate RUL predictions for validation observations.
# FR: Générer les prédictions de RUL pour les observations de validation.
y_val_pred = baseline_model.predict(X_val_model)

# EN: Calculate absolute and squared-error metrics.
# FR: Calculer les métriques d'erreur absolue et quadratique.
mae = mean_absolute_error(y_val, y_val_pred)

rmse = np.sqrt(
    mean_squared_error(y_val, y_val_pred)
)

# EN: Calculate the NASA/PHM08 asymmetric scoring function.
# FR: Calculer la fonction de score asymétrique NASA/PHM08.
errors = y_val_pred - y_val

phm08_score = np.sum(
    np.where(
        errors < 0,
        np.exp(-errors / 13) - 1,
        np.exp(errors / 10) - 1
    )
)

print("=== XGBOOST BASELINE RESULTS ===")

print(f"\nMAE: {mae:.3f} cycles")
print(f"RMSE: {rmse:.3f} cycles")
print(f"NASA/PHM08 score: {phm08_score:.3f}")

print("\nPrediction range:")
print(f"Minimum predicted RUL: {y_val_pred.min():.3f}")
print(f"Maximum predicted RUL: {y_val_pred.max():.3f}")

print("\nActual RUL range:")
print(f"Minimum actual RUL: {y_val.min():.3f}")
print(f"Maximum actual RUL: {y_val.max():.3f}")
```

```python
# STEP 12C - Baseline prediction diagnostics
# EN: Quantify prediction errors, negative predictions, and large-error cases.
# FR: Quantifier les erreurs de prédiction, les prédictions négatives et les grandes erreurs.

errors = y_val_pred - y_val
absolute_errors = np.abs(errors)

negative_predictions = y_val_pred < 0
large_errors = absolute_errors > 50

print("=== BASELINE PREDICTION DIAGNOSTICS ===")

print(f"\nTotal validation predictions: {len(y_val_pred)}")

print(
    f"Negative predictions: "
    f"{negative_predictions.sum()} "
    f"({negative_predictions.mean() * 100:.2f}%)"
)

print(
    f"Errors > 50 cycles: "
    f"{large_errors.sum()} "
    f"({large_errors.mean() * 100:.2f}%)"
)

print(
    f"Errors > 100 cycles: "
    f"{(absolute_errors > 100).sum()} "
    f"({(absolute_errors > 100).mean() * 100:.2f}%)"
)

print("\nError statistics:")
print(
    pd.Series(absolute_errors)
    .describe()
)

print("\nPrediction statistics:")
print(
    pd.Series(y_val_pred)
    .describe()
)
```

```python
# EN: Aggregate XGBoost importance by feature family.
# FR: Regrouper l'importance XGBoost par famille de variables.

def feature_family(feature):
    # EN: Identify the family of each feature.
    # FR: Identifier la famille de chaque variable.
    
    if feature == "cycle_age":
        return "cycle_age"
    elif "_rollmean_5" in feature:
        return "rolling_mean_5"
    elif "_rollstd_5" in feature:
        return "rolling_std_5"
    elif "_diff_1" in feature:
        return "difference_1"
    elif "_trend_10" in feature:
        return "trend_10"
    elif feature.startswith("sensor_"):
        return "raw_sensor"
    elif feature.startswith("setting_"):
        return "operating_setting"
    else:
        return "other"


feature_importance["family"] = (
    feature_importance["feature"].apply(feature_family)
)

family_importance = (
    feature_importance
    .groupby("family", as_index=False)["importance"]
    .sum()
    .sort_values("importance", ascending=False)
    .reset_index(drop=True)
)

family_importance["importance_pct"] = (
    family_importance["importance"] * 100
)

family_importance
```

```python
# EN: Compare PHM08 penalty contributions by actual-RUL region.
# FR: Comparer les contributions à la pénalité PHM08 selon les régions de RUL réel.

def phm08_contribution(errors):
    # EN: Calculate the PHM08 contribution of each prediction.
    # FR: Calculer la contribution PHM08 de chaque prédiction.
    return np.where(
        errors < 0,
        np.exp(-errors / 13) - 1,
        np.exp(errors / 10) - 1
    )


error_analysis["phm08_contribution"] = phm08_contribution(
    error_analysis["error"]
)

error_analysis_lgbm["phm08_contribution"] = phm08_contribution(
    error_analysis_lgbm["error"]
)

phm_regions_xgb = (
    error_analysis
    .groupby("RUL_region", observed=False)
    .agg(
        PHM08=("phm08_contribution", "sum")
    )
    .reset_index()
)

phm_regions_lgbm = (
    error_analysis_lgbm
    .groupby("RUL_region", observed=False)
    .agg(
        PHM08=("phm08_contribution", "sum")
    )
    .reset_index()
)

phm_comparison = phm_regions_xgb.copy()

phm_comparison = phm_comparison.rename(
    columns={"PHM08": "XGBoost_PHM08"}
)

phm_comparison["LightGBM_PHM08"] = (
    phm_regions_lgbm["PHM08"]
)

phm_comparison["difference"] = (
    phm_comparison["LightGBM_PHM08"]
    - phm_comparison["XGBoost_PHM08"]
)

phm_comparison
```

```python
# EN: Add longer-horizon rolling means to test whether longer temporal context
#     improves degradation representation.
# FR: Ajouter des moyennes mobiles à plus long horizon pour tester si un contexte
#     temporel plus long améliore la représentation de la dégradation.

long_feature_df = feature_df.copy()

long_windows = [10, 20]

for sensor in variable_sensors:
    for window in long_windows:
        long_feature_df[f"{sensor}_rollmean_{window}"] = (
            long_feature_df
            .groupby("unit_id")[sensor]
            .transform(
                lambda x: x.rolling(
                    window=window,
                    min_periods=window
                ).mean()
            )
        )

print("Original feature_df shape:", feature_df.shape)
print("Extended feature_df shape:", long_feature_df.shape)

new_features = [
    col for col in long_feature_df.columns
    if col not in feature_df.columns
]

print("New features:", len(new_features))
print("Expected new features:", len(variable_sensors) * len(long_windows))
print("Missing values in new features:",
      long_feature_df[new_features].isna().sum().sum())
```

# 10. Feature Engineering
# 10. Ingénierie des variables

## 10.1 Objective
## 10.1 Objectif

**EN:**  
The objective of feature engineering is to transform the raw multivariate engine telemetry into predictive variables that capture degradation-related temporal information while preserving causal ordering and avoiding data leakage.

For Aerohealth, the feature design focuses on three complementary sources of information:

1. Current sensor state
2. Short-term temporal behavior
3. Engine progression through its observed trajectory

**FR :**  
L'objectif de l'ingénierie des variables est de transformer les données brutes de télémétrie multivariée du moteur en variables prédictives capables de représenter les informations temporelles liées à la dégradation, tout en respectant l'ordre causal des observations et en évitant les fuites de données.

Pour Aerohealth, la conception des variables repose sur trois sources complémentaires d'information :

1. L'état actuel des capteurs
2. Le comportement temporel à court terme
3. La progression du moteur dans sa trajectoire observée

---

## 10.2 Design Principle: Causal Temporal Features
## 10.2 Principe de conception : variables temporelles causales

**EN:**  
All temporal features are constructed using the current observation and, where applicable, previous observations from the same engine.

Future observations are never used to construct a feature for the current cycle.

This preserves the temporal structure of the prognostics problem and makes the feature representation compatible with deployment on a progressively observed engine trajectory.

**FR :**  
Toutes les variables temporelles sont construites à partir de l'observation courante et, lorsque cela est applicable, des observations précédentes du même moteur.

Aucune observation future n'est utilisée pour construire une variable correspondant au cycle courant.

Cela permet de préserver la structure temporelle du problème de pronostic et de rendre la représentation des variables compatible avec une utilisation sur une trajectoire moteur observée progressivement.

---

## 10.3 Variable Selection
## 10.3 Sélection des variables

**EN:**  
The FD001 training data contains 21 sensor measurements. Exploratory analysis identified 15 sensors with non-constant values.

The six constant sensors were not used for temporal feature generation because they contain no variation within the FD001 training dataset.

The 15 variable sensors were therefore used to construct the temporal features.

**FR :**  
Les données d'entraînement FD001 contiennent 21 mesures de capteurs. L'analyse exploratoire a identifié 15 capteurs présentant des valeurs non constantes.

Les six capteurs constants n'ont pas été utilisés pour générer les variables temporelles, car ils ne présentent aucune variation dans le jeu de données d'entraînement FD001.

Les 15 capteurs variables ont donc été utilisés pour construire les variables temporelles.

---

## 10.4 Five-Cycle Rolling Mean
## 10.4 Moyenne mobile sur cinq cycles

**EN:**  
For each variable sensor, a five-cycle rolling mean was calculated.

The rolling mean smooths short-term fluctuations and provides an estimate of the recent sensor state.

The window is causal: only the current observation and preceding observations are considered.

At the beginning of an engine trajectory, the feature remains missing until enough observations are available.

**FR :**  
Pour chaque capteur variable, une moyenne mobile sur cinq cycles a été calculée.

Cette moyenne mobile permet de réduire les fluctuations à court terme et de représenter l'état récent du capteur.

La fenêtre est causale : seules l'observation courante et les observations précédentes sont utilisées.

Au début de la trajectoire d'un moteur, la variable reste manquante jusqu'à ce qu'un nombre suffisant d'observations soit disponible.

---

## 10.5 Five-Cycle Rolling Standard Deviation
## 10.5 Écart-type mobile sur cinq cycles

**EN:**  
A five-cycle rolling standard deviation was calculated for each variable sensor.

This feature captures short-term variability around the local sensor state.

It complements the rolling mean: the rolling mean represents the recent level, while the rolling standard deviation represents recent local variability.

**FR :**  
Un écart-type mobile sur cinq cycles a été calculé pour chaque capteur variable.

Cette variable permet de représenter la variabilité à court terme autour de l'état local du capteur.

Elle complète la moyenne mobile : la moyenne mobile représente le niveau récent du signal, tandis que l'écart-type mobile représente sa variabilité locale récente.

---

## 10.6 One-Cycle Difference
## 10.6 Différence entre deux cycles successifs

**EN:**  
For each variable sensor, a one-cycle difference was calculated:

`difference_1(t) = sensor(t) - sensor(t-1)`

This feature represents the short-term change in the sensor measurement.

The first observation of each engine has no previous observation and therefore produces a structural missing value.

**FR :**  
Pour chaque capteur variable, une différence entre deux cycles successifs a été calculée :

`difference_1(t) = sensor(t) - sensor(t-1)`

Cette variable représente la variation à court terme de la mesure du capteur.

La première observation de chaque moteur ne possède pas d'observation précédente et produit donc une valeur manquante structurelle.

---

## 10.7 Ten-Cycle Trend
## 10.7 Tendance sur dix cycles

**EN:**  
A ten-cycle rolling mean of the one-cycle difference was calculated for each variable sensor.

This feature is intended to capture a smoother local trend in sensor evolution rather than the instantaneous change between two cycles.

It remains causal because it uses only current and previous differences.

**FR :**  
Une moyenne mobile sur dix cycles des différences entre cycles successifs a été calculée pour chaque capteur variable.

Cette variable vise à représenter une tendance locale plus stable de l'évolution du capteur plutôt que la variation instantanée entre deux cycles.

Elle reste causale car elle utilise uniquement les différences courantes et précédentes.

---

## 10.8 Engine Age
## 10.8 Âge du moteur

**EN:**  
The current cycle was retained as a temporal feature under the name `cycle_age`.

This variable represents the engine's current progression in cycles.

Unlike the exploratory `relative_life` variable used during EDA, `cycle_age` does not require knowledge of the engine's final observed cycle.

Therefore, it can be constructed consistently for progressively observed trajectories.

**FR :**  
Le cycle courant a été conservé comme variable temporelle sous le nom `cycle_age`.

Cette variable représente la progression actuelle du moteur en nombre de cycles.

Contrairement à la variable exploratoire `relative_life` utilisée pendant l'EDA, `cycle_age` ne nécessite pas de connaître le dernier cycle observé du moteur.

Elle peut donc être construite de manière cohérente sur des trajectoires observées progressivement.

---

## 10.9 Relative-Life Leakage Consideration
## 10.9 Risque de fuite avec la durée de vie relative

**EN:**  
During exploratory analysis, the following coordinate was investigated:

`relative_life = cycle / maximum_cycle_of_engine`

This variable was useful for analyzing sensor behavior relative to the engine trajectory.

However, it was not retained as a predictive feature.

The calculation requires the maximum observed cycle of the engine. For a truncated test trajectory, this quantity is not known at the current prediction time.

Using it as a predictive feature would therefore introduce information that would not be available in the intended prediction setting.

**Decision:** `relative_life` is exploratory only and is excluded from the predictive feature matrix.

**FR :**  
Pendant l'analyse exploratoire, la coordonnée suivante a été étudiée :

`relative_life = cycle / maximum_cycle_of_engine`

Cette variable était utile pour analyser le comportement des capteurs relativement à la trajectoire du moteur.

Cependant, elle n'a pas été conservée comme variable prédictive.

Son calcul nécessite de connaître le cycle maximal observé du moteur. Pour une trajectoire de test tronquée, cette information n'est pas connue au moment de la prédiction.

Son utilisation comme variable prédictive introduirait donc une information qui ne serait pas disponible dans le contexte réel de prédiction.

**Décision :** `relative_life` reste exclusivement exploratoire et est exclue de la matrice des variables prédictives.

# 17. Asymmetric Regression Objective
# 17. Objectif de régression asymétrique

## 17.1 Motivation
## 17.1 Motivation

**EN:**  
The reference LightGBM model was evaluated using MAE, RMSE, and the NASA/PHM08 scoring function.

The PHM08 score is asymmetric: prediction errors in the overprediction direction receive a stronger exponential penalty than errors of the same magnitude in the underprediction direction.

The previous experiments also showed systematic RUL-dependent prediction bias.

A quantile-regression objective was therefore tested as a controlled experiment to determine whether shifting the model's prediction behavior could reduce costly overprediction errors.

**FR :**  
Le modèle LightGBM de référence a été évalué avec la MAE, la RMSE et la fonction de score NASA/PHM08.

Le score PHM08 est asymétrique : les erreurs de prédiction dans le sens de la surestimation reçoivent une pénalité exponentielle plus forte que les erreurs de même amplitude dans le sens de la sous-estimation.

Les expériences précédentes ont également montré un biais de prédiction dépendant de la région du RUL.

Un objectif de régression quantile a donc été testé dans une expérience contrôlée afin de déterminer si une modification du comportement des prédictions pouvait réduire les erreurs coûteuses de surestimation.

## 17.5 Interprétation du profil d'erreur

**EN:**  
The error plot shows that the α=0.4 model systematically shifts predictions downward relative to the reference behavior.

Cependant, cette amélioration ne signifie pas que les erreurs de prédiction ont été éliminées de manière générale. Une partie importante de la distribution des erreurs s'est déplacée vers le sens de la sous-estimation.

Par exemple, les plus grandes contributions au PHM08 sont désormais principalement associées à des sous-estimations lorsque le RUL réel est élevé.

Cela montre que la réduction du PHM08 est liée à une modification de la **direction et de l'amplitude de la distribution des erreurs**, et non simplement à une réduction de toutes les erreurs.

```python
# EN: Extract one final RUL prediction per test engine at its last observed cycle.
# FR: Extraire une prédiction finale de RUL par moteur au dernier cycle observé.

test_prediction_df = test_df[
    ["unit_id", "cycle"]
].copy()

test_prediction_df["predicted_RUL"] = y_test_pred_all

final_test_predictions = (
    test_prediction_df
    .sort_values(["unit_id", "cycle"])
    .groupby("unit_id", as_index=False)
    .tail(1)
    .sort_values("unit_id")
    .reset_index(drop=True)
)

print("Final engine-level predictions:", len(final_test_predictions))
print(
    "All 100 test engines represented:",
    final_test_predictions["unit_id"].nunique() == 100
)
print(
    "Maximum cycle per engine correctly selected:",
    all(
        final_test_predictions["cycle"].values
        == test_df.groupby("unit_id")["cycle"].max().sort_index().values
    )
)

print("\nFirst 10 final predictions:")
print(final_test_predictions.head(10).to_string(index=False))

print("\nPrediction range:",
      float(final_test_predictions["predicted_RUL"].min()),
      "to",
      float(final_test_predictions["predicted_RUL"].max()))
```

```python
# EN: Evaluate the final Aerohealth model on the official FD001 test targets.
# FR: Évaluer le modèle final Aerohealth sur les cibles officielles de test FD001.

from sklearn.metrics import mean_absolute_error, mean_squared_error

# EN: Align predictions and official targets by engine ID.
# FR: Aligner les prédictions et les cibles officielles par identifiant moteur.

final_test_eval = final_test_predictions.merge(
    test_rul.assign(unit_id=np.arange(1, len(test_rul) + 1)),
    on="unit_id",
    how="inner"
)

y_test_true = final_test_eval["RUL"].to_numpy()
y_test_pred = final_test_eval["predicted_RUL"].to_numpy()

# EN: Calculate standard regression metrics.
# FR: Calculer les métriques standards de régression.

test_mae = mean_absolute_error(y_test_true, y_test_pred)

test_rmse = np.sqrt(
    mean_squared_error(y_test_true, y_test_pred)
)

# EN: Calculate the NASA/PHM08 asymmetric scoring function.
# FR: Calculer la fonction de score asymétrique NASA/PHM08.

test_errors = y_test_pred - y_test_true

test_phm08 = np.sum(
    np.where(
        test_errors < 0,
        np.exp(-test_errors / 13) - 1,
        np.exp(test_errors / 10) - 1
    )
)

# EN: Additional diagnostics.
# FR: Diagnostics supplémentaires.

absolute_errors = np.abs(test_errors)

print("Official FD001 test evaluation")
print("=" * 40)

print("Engines evaluated:", len(final_test_eval))
print("MAE:", float(test_mae))
print("RMSE:", float(test_rmse))
print("NASA/PHM08 score:", float(test_phm08))

print("\nError diagnostics")
print("-" * 40)
print("Mean error (prediction - actual):", float(test_errors.mean()))
print("Median error:", float(np.median(test_errors)))
print("Maximum absolute error:", float(absolute_errors.max()))
print("Errors > 25 cycles:", int((absolute_errors > 25).sum()))
print("Errors > 50 cycles:", int((absolute_errors > 50).sum()))
print("Errors > 100 cycles:", int((absolute_errors > 100).sum()))
print("Negative predictions:", int((y_test_pred < 0).sum()))

print("\nPrediction / target statistics")
print("-" * 40)
print("Mean prediction:", float(y_test_pred.mean()))
print("Mean actual RUL:", float(y_test_true.mean()))
print("Median prediction:", float(np.median(y_test_pred)))
print("Median actual RUL:", float(np.median(y_test_true)))
```
